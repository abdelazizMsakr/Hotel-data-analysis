### Replacing Null values

```python
df['children'] = df['children'].fillna(0).astype(int)
```

```python
df['company'] = df['company'].fillna('N/A')
df['agent'] = df['agent'].fillna('N/A')
df['country'] = df['country'].fillna('N/A')
```

### Correlation between columns 

```python
numeric_df = df.select_dtypes( include = 'number')
corr = numeric_df.corr()

# Using PLt :
#plt.imshow(corr, cmap = 'coolwarm', aspect = 'auto', interpolation = 'nearest')
#plt.show()

# USing sns :
plt.figure(figsize = (16,9))
sns.heatmap(corr, annot = True, cmap = 'coolwarm')
plt.show()
```

## Analysis Questions :
### 1) What is the overall cancellation rate? 

```python
Cancellation_Rate = df["is_canceled"].mean().round(4) * 100
print(f'Cancellation Rate is {Cancellation_Rate} %')
```

### 2) Which hotel type has a higher cancellation rate?

```python
Cancellation_Hotel_Rate = df.groupby('hotel')['is_canceled'].mean().round(1)
plt.pie(Cancellation_Hotel_Rate, labels = Cancellation_Hotel_Rate.index, autopct = '%1.1f')
plt.show()
```

### 3) What are the top 5 market segments by booking volume?

```python
Market_segment_booking = df.groupby('market_segment')['market_segment'].count().sort_values(ascending = False).head()
bars = plt.bar(Market_segment_booking.index, Market_segment_booking)
plt.bar_label(bars, fmt = '%d', padding = 3)
plt.show()
```

### 4) Which market segment has the highest cancellation rate? 

```python
Market_Segmant_Cancelation = df.groupby('market_segment')['is_canceled'].mean() * 100
Market_Segmant_Cancelation = Market_Segmant_Cancelation.sort_values( ascending = False).head(1)

print(f'The Market Segment with highest cancellation Rate is {Market_Segmant_Cancelation.index[0]} = {Market_Segmant_Cancelation.iloc[0]}%')
```

### 5) Which distribution channel is most risky? 

```python
Distribution_Channel_Cancelation = df.groupby(['distribution_channel','is_canceled'])['is_canceled'].value_counts().sort_values(ascending = False)
Distribution_Channel_Cancelation
```

### 6) Is deposit type a strong predictor of cancellation?

```python
Deposite_Type_Cancellation = df.groupby('deposit_type')['is_canceled'].mean() * 100
plt.pie(Deposite_Type_Cancellation, labels = Deposite_Type_Cancellation.index, autopct = '%1.1f')
plt.show()
```

### 7) How does lead time affect cancellation?

```python
Lead_time_Cancelation_df = df[['lead_time','is_canceled']]

corr = Lead_time_Cancelation_df.corr()
sns.heatmap(corr, annot = True, cmap = 'coolwarm')
plt.show()
```

```python
bins = [-1,7,15,60,90,180,365,737]
labels = ['0-7','8-15','16-60','61-90','91-180','181-365','+365']
df['LT_bins'] = pd.cut(df['lead_time'], bins=bins, labels=labels)
LT_bins_Cancelation = df.groupby('LT_bins')['is_canceled'].mean().round(4).sort_values(ascending = False) * 100
```

```python
bars = plt.bar(LT_bins_Cancelation.index, LT_bins_Cancelation)
plt.bar_label(bars, fmt = '%d', padding = 3)
plt.show()
```

```python
LT_bins_Cancelation2 = df.groupby('LT_bins')['is_canceled'].sum().round(4).sort_values(ascending = False) * 100
plt.pie(LT_bins_Cancelation2, labels = LT_bins_Cancelation2.index, autopct = '%d')
plt.show()
```

### 8) Is ADR higher for canceled bookings?

```python
Adr_Cancelation_df = df[['adr','is_canceled']]
corr_Adr_Cancelation = Adr_Cancelation_df.corr()
sns.heatmap(corr_Adr_Cancelation, annot = True, cmap = 'coolwarm')
plt.show()
```

### 9) Which months have the highest cancellation rates?

```python
# Sorting months :
month_order = ['January', 'February', 'March', 'April', 'May', 'June','July', 'August', 'September', 'October', 'November', 'December']
df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories = month_order, ordered = True)

Month_Cancelation = df.groupby('arrival_date_month')['is_canceled'].sum()

# Line chart with plt :
plt.figure(figsize = (16,6))
plt.plot(Month_Cancelation.index, Month_Cancelation, marker = 'o', linestyle = '-')
plt.show()

# Using sns :
```

### 10) Does having children affect cancellation?

```python
df['children'] = df['children'].astype(int)
df['Have_Child'] = df['children'].apply(lambda x : 'Have Children' if x > 0 else 'No Children')

Children_Cancelation = df.groupby('Have_Child')['is_canceled'].mean().round(4) * 100
plt.pie(Children_Cancelation, labels = Children_Cancelation.index, autopct = '%1.1f')
plt.show()
```

### 11) Do repeated guests cancel less?

```python
Repeated_Guests_Cancelation = df['is_repeated_guest'].mean().round(4) * 100
print(f'The Cancelation Rate for Reapeated Guests is {Repeated_Guests_Cancelation} %')
```

### 12) Does being on a waiting list increase cancellation?

```python
Waiting_list_labels = ['-1:7','8:16','17:30','31:60','61:120','121:240','+240']
Waiting_bins = [-1,7,16,30,60,120,240,391]
df['Waiting_Days_Range'] = pd.cut(df['days_in_waiting_list'], bins = Waiting_bins, labels = Waiting_list_labels)

Waiting_List_Cancelation = df.groupby('Waiting_Days_Range')['is_canceled'].mean().round(4).sort_values(ascending = False) * 100
Waiting_List_Cancelation
```

### 13) Are booking changes correlated with cancellation?

```python
Booking_Changes_Labels = ['No Changes','1-5','6-10','11-20','+20']
Changes_bins = [0,1,5,10,20,21]
df['Booking_Changes_Range'] = pd.cut(df['booking_changes'], bins = Changes_bins, labels = Booking_Changes_Labels)

Booking_Changes_Cancelation = df.groupby('Booking_Changes_Range')['is_canceled'].mean().round(4).sort_values(ascending = False) * 100
Booking_Changes_Cancelation
```

```python
Booking_Cancelation = df[['booking_changes','is_canceled']]
Corr_Booking_Cancelation = Booking_Cancelation.corr()
sns.heatmap(Corr_Booking_Cancelation, annot = True, cmap = 'coolwarm')
plt.show()
```

### 14) Do special requests reduce cancellation?

```python
df_SR_Cancelation = df[['total_of_special_requests','is_canceled']]
SpecialRequests_Cancelation_Corr = df_SR_Cancelation.corr()
sns.heatmap(SpecialRequests_Cancelation_Corr, annot = True, cmap = 'coolwarm')
plt.show()
```

### 15) Does room mismatch increase cancellation?

```python
df['Room_Mismatch'] = df.apply(lambda x : 'Match' if x['reserved_room_type'] == x['assigned_room_type'] else 'Mismatch', axis =1)

Room_matching_cancelation = df.groupby('Room_Mismatch')['is_canceled'].mean().round(4) * 100
plt.pie(Room_matching_cancelation, labels = Room_matching_cancelation.index, autopct = '%1.1f')
plt.show()
```
